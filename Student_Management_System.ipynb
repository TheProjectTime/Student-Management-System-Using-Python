```python
import csv
import os

FILE_NAME = "students.csv"


# Create CSV file if it does not exist
def initialize_file():
    if not os.path.exists(FILE_NAME):
        with open(FILE_NAME, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(["Roll Number", "Name", "Marks"])


# Add a new student
def add_student():
    roll_no = input("Enter Roll Number: ")
    name = input("Enter Student Name: ")
    marks = input("Enter Marks: ")

    # Check if roll number already exists
    with open(FILE_NAME, "r", newline="") as file:
        reader = csv.DictReader(file)

        for student in reader:
            if student["Roll Number"] == roll_no:
                print("Student with this Roll Number already exists!")
                return

    # Save student to CSV file
    with open(FILE_NAME, "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([roll_no, name, marks])

    print("Student added successfully!")


# Display all students
def display_students():
    with open(FILE_NAME, "r", newline="") as file:
        reader = csv.DictReader(file)

        students = list(reader)

        if not students:
            print("No student records found.")
            return

        print("\n--- Student Records ---")
        print(f"{'Roll No.':<12}{'Name':<25}{'Marks':<10}")
        print("-" * 47)

        for student in students:
            print(
                f"{student['Roll Number']:<12}"
                f"{student['Name']:<25}"
                f"{student['Marks']:<10}"
            )


# Search for a student
def search_student():
    roll_no = input("Enter Roll Number to search: ")

    with open(FILE_NAME, "r", newline="") as file:
        reader = csv.DictReader(file)

        for student in reader:
            if student["Roll Number"] == roll_no:
                print("\nStudent Found!")
                print("Roll Number:", student["Roll Number"])
                print("Name:", student["Name"])
                print("Marks:", student["Marks"])
                return

    print("Student not found.")


# Delete a student
def delete_student():
    roll_no = input("Enter Roll Number to delete: ")

    students = []
    found = False

    # Read all students
    with open(FILE_NAME, "r", newline="") as file:
        reader = csv.DictReader(file)

        for student in reader:
            if student["Roll Number"] == roll_no:
                found = True
            else:
                students.append(student)

    if not found:
        print("Student not found.")
        return

    # Rewrite the CSV file without the deleted student
    with open(FILE_NAME, "w", newline="") as file:
        fieldnames = ["Roll Number", "Name", "Marks"]
        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(students)

    print("Student deleted successfully!")


# Main program
def main():
    initialize_file()

    while True:
        print("\n========== STUDENT MANAGEMENT SYSTEM ==========")
        print("1. Add Student")
        print("2. Display Students")
        print("3. Search Student")
        print("4. Delete Student")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == "1":
            add_student()

        elif choice == "2":
            display_students()

        elif choice == "3":
            search_student()

        elif choice == "4":
            delete_student()

        elif choice == "5":
            print("Thank you for using Student Management System!")
            break

        else:
            print("Invalid choice! Please try again.")


# Start the program
if __name__ == "__main__":
    main()
```
